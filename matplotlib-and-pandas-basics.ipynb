# Matplotlib and Pandas Basics
Practice notebook: basic plots with Matplotlib and getting started with Pandas.

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## Line plot

```python
#line plot (1,2) & (3,4)
xpoint = np.array([1,3])
ypoint = np.array([2,4])
plt.plot(xpoint,ypoint)
plt.show()
```

```python
cgpa = np.array ([6, 6.5, 7, 7.8, 8, 8.9, 9, 9.5, 10 ])
package = np.array ([4, 4.5, 6, 6.5, 7, 7.8, 8, 8.9 , 10])
font1 = {'family':'serif','color':'blue','size':20}
font2 = {'family':'serif','color':'darkred','size':15}
plt. plot (cgpa,package,"*:g",ms=15,mec="red")
plt.xlabel ("cgpa")
plt.ylabel ("package")
plt.title ("cgpa vs package")
plt.grid(color="blue",linestyle = ":",linewidth = 1.5)
plt.show ()
```

## Bar plots

```python
# bar plot -> it is work on numerical as well as categorical data
fruit = np.array(["Apple","Banana", "orange","watermelon"])
quantity = np.array([100,200,150,120])
color = np.array(["Red","yellow","orange","green"])
plt.bar (fruit,quantity,color = color)
plt.show()
```

```python
# bar plot -> it is work on numerical as well as categorical data

fruit = np.array(["Apple","Banana", "orange","watermelon"])
quantity = np.array([100,200,150,120])
color = np.array(["Red","yellow","orange","green"])
plt.barh (fruit,quantity,color = color)
plt.show()
```

## Pie chart

```python
# pie chart -> it is work on numerical as well as categorical data
fruit_percentage = np.array([20,30,35,20])
fruit = np.array(["Apple","Banana", "orange","watermelon"])
plt.pie(fruit_percentage,labels = fruit)
plt.legend(title = "fruit")
plt.show()
```

## Scatter plot

```python
# scatter pot ->  this plot represent data in point form
age_of_car = np.array([4,5,6,7,8,9,10])
speed_of_car = np.array([150,140,130,120,110,100,90])
plt.scatter(age_of_car,speed_of_car)
plt.show()
```

## Histogram

```python
x = np.random.randint(40,100,(1,200))
x = x.flatten()
print(x)
```

```python
plt.hist(x,bins = 20)
plt.xlabel("Data")
plt.ylabel("frequency")
plt.show
```

## Cosine similarity

```python
from numpy.linalg import norm

A = np.array([1, 2, 3, 4])
B = np.array([1, 2, 4, 5])
cosine = np.dot(A, B) / (norm(A) * norm(B))
print(cosine)
```
